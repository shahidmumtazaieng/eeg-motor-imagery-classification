# Signal analysis

Inspect sensor locations and the mean EEG power spectral density after the project preprocessing steps. Visual output is exploratory and should not be mistaken for a statistical test.

```python
import matplotlib.pyplot as plt
import numpy as np
from src.preprocessing import load_subject_epochs

epochs = load_subject_epochs(1)
epochs.average().plot(picks=['C3', 'C4', 'Cz'], show=True)
epochs.compute_psd(fmin=1, fmax=45).plot(average=True, show=True)
epochs['left_hand'].average().plot_topomap(times=[1.0, 2.0, 3.0], ch_type='eeg', show=True)
tfr = epochs['left_hand'].compute_tfr(
    method='morlet', freqs=np.arange(6, 31), n_cycles=np.arange(6, 31) / 2,
    return_itc=False, average=True, decim=3, verbose='ERROR'
)
tfr.plot(picks=['C3', 'C4'], show=True)
plt.show()
```
