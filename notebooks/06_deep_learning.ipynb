# Deep learning: EEGNet-style model

This participant-level demonstration trains only on training trials. Use the subject-independent experiment in `05_classical_ml.ipynb` or `src.train` for the main held-out-participant comparison.

```python
import numpy as np
from sklearn.model_selection import StratifiedShuffleSplit
from src.data import load_dataset
from src.models import EEGNet

X, y, groups, channels, sfreq = load_dataset([1])
train, test = next(StratifiedShuffleSplit(
    n_splits=1, test_size=0.25, random_state=42
).split(X, y))
predicted, positive_score = EEGNet.fit_predict(
    X[train], y[train], X[test], epochs=30
)
print(f"test accuracy: {np.mean(predicted == y[test]):.3f}")
```
